```python
df_both = pd.concat([df_sevall, df_eightall])
```

```python
g = sns.catplot(
    data=df_both, kind="bar",
    x="model_name", y="test_acc", col='resolution', hue='model_name',
    errorbar="sd", palette="colorblind", alpha=0.6, height=6
)

# Function to add stripplot to each subplot
def add_stripplot(x, y, **kwargs):
    ax = plt.gca()
    data = kwargs.pop("data")
    sns.stripplot(x=x, y=y, data=data, ax=ax, dodge=True, alpha=0.8, linewidth=1)

# Apply the stripplot function to each subplot
g.map_dataframe(add_stripplot, "model_name", "test_acc")

# Adjust x-axis labels and add horizontal line
for ax in g.axes.flat:
    ax.tick_params(axis='x', rotation=90)
    ax.axhline(y=11, color="0.4", linestyle="--", zorder=0)
    ax.set_ylim(0, 100)

# Set the title
g.fig.suptitle("7C and 8C. dropout 0.2a. 60 epochs. LR = 1e-4. MSE", y=1.05)

# Show the plot
plt.show()
```
