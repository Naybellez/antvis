```python


def merging2onedf(folder, **kwargs):
    # for reading in csvs.
    # turning to pandas df
    # option to add additional cols
    
    used_id_list = []
    df_list = []
    #alphabet = ["B", "C", "D", "E", "F"] 
    for file in os.listdir(folder):
        if 'csv' in file:
            #if file[5:8] in used_id_list: # Check if the vid ID has already been used
            #    fileID = file[5:8]+"B"
            #    print(fileID) # show which IDs this happened to
            #else:
            if file[0] == 'A' or file[0]== 'B':
                additional = file[0]
            else:
                additional = ""
            fileID = file[5:8]+additional
            df = pd.read_csv(folder+file)
            df.insert(loc=0, column='ID', value=fileID)
            df['ID'].fillna(fileID, inplace=True)
            for key, val in kwargs.items():
                df.insert(loc=0, column=key, value=val)
                # fill all rows with the same val
                df[key].fillna(val, inplace=True)
            df_list.append(df)

            used_id_list.append(fileID)

                
    df_all = pd.concat(df_list)
    return df_all

```

```python
df_jul18 = merging2onedf(folder_jul18, Date = "Jul-18")
df_jul19 = merging2onedf(folder_jul19, Date = "Jul-19")

df_aug09 = merging2onedf(folder_aug09, Date ="Aug-09")
df_aug12 = merging2onedf(folder_aug12, Date ="Aug-12")
df_aug13 = merging2onedf(folder_aug13, Date ="Aug-13")
df_aug14 = merging2onedf(folder_aug14, Date ="Aug-14")
df_aug15 = merging2onedf(folder_aug15, Date ="Aug-15")
df_aug16 = merging2onedf(folder_aug16, Date ="Aug-16")

df_aug21 = merging2onedf(folder_aug21, Date ="Aug-21")
df_aug23 = merging2onedf(folder_aug23, Date ="Aug-23")
df_aug27 = merging2onedf(folder_aug27, Date ="Aug-27")
df_aug28 = merging2onedf(folder_aug28, Date ="Aug-28")
df_aug30 = merging2onedf(folder_aug30, Date ="Aug-30")
df_sep09 = merging2onedf(folder_sep09, Date ="Sep-09")
df_sep10 = merging2onedf(folder_sep10, Date ="Sep-10")
df_sep11 = merging2onedf(fodler_sep11, Date ="Sep-11")
df_sep20 = merging2onedf(folder_sep20, Date ="Sep-20")
df_sep23 = merging2onedf(folder_sep23, Date ="Sep-23")
df_sep24 = merging2onedf(folder_sep24, Date ="Sep-24")
df_sep26 = merging2onedf(folder_sep26, Date ="Sep-26")
```
