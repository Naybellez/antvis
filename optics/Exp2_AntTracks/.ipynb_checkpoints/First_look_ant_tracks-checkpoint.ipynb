## functions

```python
def convert_pkl2df(filename):
    with open(filename, 'rb') as f:
        filepkl = pickle.load(f)
    frames = []
    for key, val in filepkl.items():
        #print('key: ', key,'val: ',val)
        if 'frame' in key:
            frames.append(val)
    df_frames = pd.DataFrame(frames)
  
    df_frames['coordinates'] = df_frames['coordinates'].map(lambda x: x[0][0][0]) #[0][0][0] 
    df_frames['confidence'] = df_frames['confidence'].map(lambda x: x[0][0][0]) #[0][0][0] 
    return df_frames
```

```python
import numpy as np
import matplotlib.pyplot as plt

def create_polar_plot(df_coordinates):
    # Convert list format to separate x and y arrays
    #print(df_coordinates)
    origin = (1000-270, 270)
    x = np.array([coord[0] - origin[0] for coord in df_coordinates])
    y = np.array([coord[1] - origin[1] for coord in df_coordinates])
    #print(type(x), x)
    
    # Convert to polar coordinates
    r = np.sqrt(x**2 + y**2)  # radius
    theta = np.arctan2(y, x)  # angle in radians
    
    # Create the polar plot
    fig = plt.figure(figsize=(10, 10))
    ax = fig.add_subplot(111, projection='polar')
    
    # Plot the points
    ax.scatter(theta, r, alpha=0.5)
    
    # Customize the plot
    ax.set_title('Coordinate Points in Polar Space')
    ax.grid(True)
    
    return fig

create_polar_plot(df_frames['coordinates'])
    
    
# Usage example:
# fig = create_polar_plot(df_frames['coordinates'])
# plt.show()
```
