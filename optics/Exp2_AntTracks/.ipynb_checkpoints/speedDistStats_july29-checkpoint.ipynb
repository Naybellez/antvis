```python
def chunk_path(x, y, chunk=5):
    out = []
    hold = []
    for idx, j in enumerate(x):
        if len(hold) <= chunk:
            hold.append((x[idx], y[idx]))
        if len(hold) == chunk or idx == len(x)-1:
            inter = hold.copy()
            out.append(inter)
            hold.clear()        
        idx+=1
    return out

def get_deltachunkdist(path):
    allchunkdist = 0
    #p = 0
    chunkdists = []
    hold = []
    #pp = []
    for chunk in path: # i want the total distance travelled in a chunk
        #print(f"len of chunk  {len(chunk)}")
        lc = len(chunk)
        for i in range(0, lc-1):
            cd = np.sqrt((chunk[i-1][0] -chunk[i][0])**2 + (chunk[i-1][1] - chunk[i][1])**2) # calc dist between chunk elements
            allchunkdist += cd
            #p +=cd
            hold.append(cd)
            if len(hold) == len(chunk)-1: # if chunk has 4 elements, hold will have 3 (2-1, 3-2, 4-3)
                k = hold.copy()
                chunkdists.append(sum(k))
                #pp.append(p)
                hold.clear()
    return allchunkdist, chunkdists#, pp


def get_straightchunks(x_, y_, chunkval):
    path = chunk_path(x_, y_, chunkval)
    bee = [((chunk[-1][0]- chunk[0][0])**2 + (chunk[-1][1] - chunk[0][1])**2)**0.5 for chunk in path]
    d, dist = get_deltachunkdist(path)
    straight = [bee[i]/dist[i] for i in range(len(dist))]
    heading_t, heading_r = cart2pola(x_, y_)
    
    return straight, (heading_t, heading_r)

"""path5 = chunk_path(x_, y_, 5)#  path chunk      
        #print(f"len p5 / number of chunks {len(path5)}")
        bee5 = [((chunk[-1][0]- chunk[0][0])**2 + (chunk[-1][1] - chunk[0][1])**2)**0.5 for chunk in path5] # beeline per chunk
        # distance travelled in chunks
        d5, dist5 = get_deltachunkdist(path5) # dist travelled er chunk
        straight5 = [bee5[i]/dist5[i] for i in range(len(dist5))]
        
        #print(f"straight5 : {straight5}")"""


def circ_r(theta):
    # custom function version of the matlab circ_r() as equivilant not found in python (scipy & Astro)
    # set parameter values based on matlab function, found in link below (Accessed 080825)
    # https://github.com/circstat/circstat-matlab/blob/master/circ_r.m
    #] ADAPTED WITH CHATgpt5
    #w = np.ones(len(alpha))
    theta = np.array(theta)
    c = np.mean(np.cos(theta))
    s = np.mean(np.sin(theta))
    return np.sqrt(c**2 + s**2)

def circ_r_weighted(theta, w=None, axis=None):
    theta = np.array(theta)

    if w is None:
        w = np.ones_like(theta)
    w = np.array(w)

    sum_w = np.sum(w, axis=axis)
    c= np.sum(w * np.cos(theta), axis=axis)/ sum_w
    s = np.sum(w * np.sin(theta), axis=axis) / sum_w
    return np.sqrt(c**2 + s**2)

def trimPath(rad, big:int, smol:int):
    if max(rad) > big:                      # last datapoint when ant crosses the 29cm boundry
        distBigger29 = np.where(rad<big)
        lastpoint = distBigger29[0][-1]
    else:
        lastpoint = len(rad)
    if max(rad) <= smol:                        # first data point when ant crosses the 3cm boundry
        firstPoint = 0
    else:
        distBigger20 = np.where(rad>smol)  # returns index
        #print("rads bigger than smol (30)",distBigger20[0])
        firstPoint = distBigger20[0][0]
    return firstPoint, lastpoint


def get_chunkedPath(accumulatedDist:list, x, y, chunksize=20):
    oldr = 0
    start = 0
    dist_trav_chunk = 0
    chunkedPath2cm = []               # to contain (theta, r) at 2cm intervals
    numPoints = np.floor(accumulatedDist[-1]/chunksize) # how mancy chunks of 2cm
    newx = []
    newy = []
    #print(f" trimmed_dist{trimmed_dist}")
    for idx,p in enumerate(accumulatedDist[1:]):  # 1. what we are iterating through    2. start index
        if oldr == 0:
            oldr = p                   # get the first radius position
        # find next point 2cm away

        if abs(abs(p) - abs(oldr)) >= chunksize:          # if the difference between the current position and the last logged pos is greater or equal to 2cm (20 radians)/ our chunk size
            newx.append(x[idx])
            newy.append(y[idx])
            oldr = p
            dist_trav_chunk += (p-oldr)
        idx +=1
    return np.array(newx), np.array(newy)


def turnCol2array(df, colname:str):
    return np.array(df[colname])


def StraightnessIndex(accumlativeDist, x, y):
    total_dist_travelled = accumlativeDist[np.logical_not(np.isnan(accumlativeDist))][-1]
    beeline = ((x[-1] - x[1])**2 + (y[-1] - y[1])**2)**0.5
    return beeline / total_dist_travelled, beeline
```
