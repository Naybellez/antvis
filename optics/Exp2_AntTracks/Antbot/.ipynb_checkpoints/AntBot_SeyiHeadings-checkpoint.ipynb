## Speed

```python
def dist_speed(df):

    usedf = df.copy()
    #print(usedf.keys, '--')
    fps = 1/100
    
    _x = np.array(usedf['x'])
    _y = np.array(usedf['y'])
    #print(usedf['x'].diff())
    theta = np.arctan2(_y, _x) # radians for polar plotting

    usedf['deltadist'] = np.sqrt(usedf.loc[:,'x'].diff()**2 + usedf.loc[:,'y'].diff()**2)
    #print(usedf.head())
    usedf['speed'] = usedf.loc[:,'deltadist'] * fps
    #print(usedf.head())
    usedf['accumdist'] = usedf.loc[:,'deltadist'].cumsum()
    totalDistTravelled = usedf['accumdist'].iloc[-1]
    averageSpeed = totalDistTravelled / (int(usedf['coords'].iloc[-1]) * fps)

    beeline = ((_x[-1] - _x[1])**2 + (_y[-1] - _y[1])**2)**0.5
    straightness = beeline / totalDistTravelled

    print(f"speed  ::  MIN: {usedf['speed'].min()},  MAX: {usedf['speed'].max()}, MEAN: {np.mean(usedf['speed'])}, Median: {usedf['speed'].median()}")
    print(f"")
    print(f"")
    print(f"")
    print(f"")

dist_speed(cleanHeadings)
```

## calc dist

```python
def dist_speed(df):

    usedf = df.copy()
    #print(usedf.keys, '--')
    fps = 1/100
    
    _x = np.array(usedf['body_x_cm'])
    _y = np.array(usedf['body_y_cm'])
    #print(usedf['x'].diff())
    theta = np.arctan2(_y, _x) # radians for polar plotting

    usedf['deltadist'] = np.sqrt(usedf.loc[:,'body_x_cm'].diff()**2 + usedf.loc[:,'body_y_cm'].diff()**2)
    #print(usedf.head())
    usedf['speed'] = usedf.loc[:,'deltadist'] * fps
    #print(usedf.head())
    usedf['accumdist'] = usedf.loc[:,'deltadist'].cumsum()
    totalDistTravelled = usedf['accumdist'].iloc[-1]
    averageSpeed = totalDistTravelled / (int(usedf['frame_number'].iloc[-1]) * fps)

    beeline = ((_x[-1] - _x[1])**2 + (_y[-1] - _y[1])**2)**0.5
    straightness = beeline / totalDistTravelled

    print(f"speed  ::  MIN: {usedf['speed'].min()},  MAX: {usedf['speed'].max()}, MEAN: {np.mean(usedf['speed'])}, Median: {usedf['speed'].median()}")
    print(f"")
    print(f"")
    print(f"")
    print(f"")

dist_speed(antA050723)
```

# food source as center ref

```python
foodcenter = (795, 707)

df = antB050723.copy()
coords = zip(df['body_x'], df['body_y'])
newx = []
newy = []
for coord in coords:
    #print(f"coord x {coord[0]}   food x {foodcenter[0]}")
    #print(coord[0] - foodcenter[0])
    x = coord[0] - foodcenter[0]
    y= coord[1] - foodcenter[1]
    newx.append(x)
    newy.append(y)
    
df['new_x'] = newx
df['new_y'] = newy
```
