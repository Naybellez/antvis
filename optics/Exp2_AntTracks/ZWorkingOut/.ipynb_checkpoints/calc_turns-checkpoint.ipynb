```python
 # create lists to store one time metrics for later plotting
def countingTurns(antdf, id="ANTID"):
    straight_list = []
    speed_list = []
    dist_list = []
    framesPerSec = 1/100#25  # video recorded in 25 frames/ second
    stepsizes = [5, 10, 20, 40]#5, 10, 20, 20, 40, 80, 160, 320]
    mult_val = 1.5
    
    usedf = antdf.copy() # ant1050723.copy()
    previndex = 0
    x_ = np.array(usedf['body_x'])
    y_ = np.array(usedf['body_y'])
    #id = "ant1050723"
    turnsdict = {id : {}}
    for step in stepsizes:
        print(f"new step {step}")
        turnsdict[id].update({f'step {step}' :{}})
        x_sample = []
        y_sample = []
        
        while (previndex+step)-1 < len(usedf):
            x_sample.append((x_[previndex:previndex+step], previndex))
            y_sample.append((y_[previndex:previndex+step], previndex))
            previndex += step
        stepcount = len(x_sample)
        for i in range(stepcount):
            
            start_x = x_sample[i][0][0]
            start_y = y_sample[i][0][0]
            end_x = x_sample[i][0][-1]
            end_y = y_sample[i][0][-1]
            delta_dist = np.sqrt(np.diff(x_sample[i][0])**2 + np.diff(y_sample[i][0])**2)
            max_delta_dist = max(delta_dist)
            total_dist = delta_dist.cumsum()
            total_travelled = total_dist[-1]
            direct_dist = np.sqrt((end_x-start_x)**2 + (end_y - start_y)**2)
            print(f"ss{step} total dist travelled {total_travelled}")
            print(f"ss{step}  direct dist {direct_dist}")
            print(f"ss{step} Largest ant step  {max_delta_dist}")
    
            if total_travelled - direct_dist > (max_delta_dist*mult_val):
                turnsdict[id][f'step {step}'].update({i : {}})
                fig = plt.figure(figsize=(10, 10))
                ax = fig.add_subplot(221)
                ax1 = fig.add_subplot(222)
                bkgrnd = cv2.imread("testA_ant1_image.jpg")
                ax1.imshow(bkgrnd)
                
                scat1 = ax.scatter(x_sample[i][0], y_sample[i][0], c= x_sample[i][0],cmap='viridis')
                ax1.scatter(x_sample[i][0], y_sample[i][0], c= x_sample[i][0],cmap='viridis')
                plt.suptitle(f"ID {id}  {x_sample[i][1]} -- {x_sample[i][1]+step}  frames \n max_delta_dist*multval {max_delta_dist*mult_val}")
                legend = fig.legend(*scat1.legend_elements(), loc = "outside upper right", title='frames')
                ax.add_artist(legend)
                
                
                turnsdict[id][f'step {step}'][i].update({'x' : x_sample[i][0],'y': y_sample[i][0], 'framestart':x_sample[i][1]})
                plt.show()
        print(f"fin {step}")
        previndex = 0
    return turnsdict
        
```

# Frequency Domain Analysis

[using this geeks4geeks tut](https://www.geeksforgeeks.org/spectrum-analysis-in-python/)

```python
# fast fourier transform
"""
Google AI Overview:
1-D Fast Fourier Transform is algo to decompose a 1D signal into constituent frequencies.
"""

from scipy.signal import periodogram

x_ = np.array(ant1050723['body_x'])
y_ = np.array(ant1050723['body_y'])
#print(len(x_))
# delta_dist = np.sqrt(np.diff(x_sample[i][0])**2 + np.diff(y_sample[i][0])**2)
pola_coords = []
base_coords = []
for x, y in zip(x_, y_):
    p = cart2pola(x, y)
    pola_coords.append(p)
    base_coords.append((x, y))
    
#print(len(pola_coords))
fft_pol = np.fft.fft(a=pola_coords, n=len(pola_coords))
#n= fft_pol.size
freq_pol = np.fft.fftfreq(fft_pol.shape[-1], d = 1/100)

plt.hist(fft_pol)
plt.title("FFT Polar Coords \n plt.hist")
plt.show()

plt.plot(freq_pol, np.abs(fft_pol))
plt.title("FFT polar coords \n plt.plot")
plt.xlabel("freq (Hz)")
plt.ylabel("Amplitude")
plt.show()
"""

"""

fft_cart = np.fft.fft(a=base_coords, n=len(base_coords))
#n= fft_pol.size
freq_cart = np.fft.fftfreq(fft_cart.shape[-1], d = 1/100)

plt.hist(fft_cart)
plt.title("FFT Cartesian frequency \n plt.hist")
plt.show()

plt.plot(freq_cart)
plt.title("FFT Cartesian Coords \n plt.plot")
plt.xlabel("freq (Hz)")
plt.ylabel("Amplitude")
plt.show()





```

```python
"""
Google AI overview:
periodogram from series analysis. tool to identify dominant frequencies within a dataset
"""


pola_freaks, pola_powers = periodogram(pola_coords)


print("freak",len(pola_freaks))
print("power",len(pola_powers))
#plt.semilogy(freaks, powers)
#plt.hist(pola_powers)
plt.plot(pola_freaks)
plt.title("Pola frequencies periodogram \n plt.plot")
plt.xlabel("")
plt.ylabel("")
plt.show()

plt.hist(pola_freaks)
plt.title("Pola frequencies periodogram \n plt.hist")
plt.xlabel("")
plt.ylabel("")
plt.show()

plt.hist(pola_powers)
plt.title("Pola powers periodogram \n plt.hist")
plt.xlabel("")
plt.ylabel("")
plt.show()

plt.semilogy(pola_powers)
plt.title("Pola powers periodogram \n plt.semilogy")
plt.xlabel("")
plt.ylabel("")
plt.show()


```

```python
cart_freaks, cart_powers = periodogram(base_coords)


plt.plot(cart_freaks)
plt.title("cart frequencies \n periodogram \n plt.plot")
plt.xlabel("")
plt.ylabel("")
plt.show()

plt.hist(cart_freaks)
plt.title("cart frequencies \n periodogram \n plt.hist")
plt.xlabel("")
plt.ylabel("")
plt.show()

plt.hist(cart_powers)
plt.title("cartesian powers \n periodogram \n plt.hist")
plt.show()

plt.semilogy(cart_powers)
plt.title("cart powers \n periodogram \n plt.semilogy")
plt.xlabel("")
plt.ylabel("")
plt.show()
```

```python
"""
from scipy documentation: https://docs.scipy.org/doc/scipy/reference/generated/scipy.signal.welch.html

Welch's methods computes an estimate of the power spectral density by diving the data into overlapping segments,
computing a modified periodogram for each segment and averaging the periodograms


Semiology from matplotlib doc: https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.semilogy.html
a plot with scaling on the y-axis
"""

from scipy.signal import welch

welch_freak, welch_psd = welch(pola_coords)
plt.semilogy(welch_freak)
plt.title("welch frequency pola coords \n plot with semilogy")
plt.show()

plt.plot(welch_freak)
plt.title("welch frequency pola coords \n plot with plt.plot")
plt.show()

plt.hist(welch_freak)
plt.title("welch frequency pola coords \n plot with plt.hist")
```

```python
plt.semilogy(welch_psd)
plt.title("welch psd pola coords \n plot with semilogy")
plt.show()

plt.plot(welch_psd)
plt.title("welch psd pola coords \n plot with plt.plot")
plt.show()

plt.hist(welch_psd)
plt.title("welch psd pola coords \n plot with plt.hist")
```
