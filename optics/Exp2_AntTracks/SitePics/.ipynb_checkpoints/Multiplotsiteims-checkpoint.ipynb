```python
import matplotlib.pyplot as plt

import numpy as np

import cv2
import os
```

```python
n1Rpath = "/its/home/nn268/antvis/antvis/optics/Captains_Log/Exp2_AntTracks/SitePics/N1/rect/"
n2Rpath = "/its/home/nn268/antvis/antvis/optics/Captains_Log/Exp2_AntTracks/SitePics/N2/rect/"
```

```python
count = 0

for file in os.listdir(n1Rpath):
    if file[-3:].lower() == 'jpg':
        count+=1

print(f"there are {count} images in folder N1")


count = 0

for file in os.listdir(n2Rpath):
    if file[-3:].lower() == 'jpg':
        count+=1

print(f"there are {count} images in folder N2")
```

```python
n1N = []
n1E = []
n1S = []
n1W = []
n1release = []

for file in os.listdir(n1Rpath):
    if file[-3:].lower() == 'jpg':
        print(file)
        if 'release' in file.lower():
            n1release.append(file)
        elif '_E' in file:
            n1E.append(file)
        elif '_W' in file:
            n1W.append(file)
        elif 'N2m' in file:
            n1N.append(file)
        elif '_S' in file:
            n1S.append(file)

print("release",len(n1release))
print("N",len(n1N))
print("E",len(n1E))
print("S",len(n1S))
print("W",len(n1W))
```

```python
for im in n1S:
    plt.imshow(cv2.imread(n1Rpath+im))

    plt.show()
```

## N2 route follow

```python
n2_routefollow = []
for file in os.listdir(n2Rpath):
    if file[-3:].lower() == 'jpg':
        if 'onroute1' in file:
            n2_routefollow.append(file)
        elif 'base' in file:
            n2_routefollow.append(file)

print(len(n2_routefollow))
print(n2_routefollow)
#n2_routefollow_ = [n2_routefollow[0], n2_routefollow[2], n2_routefollow[1]]
n2_routefollow_ = [n2_routefollow[3], n2_routefollow[0], n2_routefollow[2], n2_routefollow[1]]

print(n2_routefollow_)
```

```python
n2_routefollow_[0]
```

```python
cols = 1
rows = 4

fig, axes = plt.subplots(rows,cols) 
fig.suptitle("Nest 2. On Route.")
axes[0].imshow(cv2.imread(n2Rpath+n2_routefollow_[0]))
axes[0].axis(False)
axes[0].set_ylabel("Nest")
axes[1].imshow(cv2.imread(n2Rpath+n2_routefollow_[1]))
axes[1].axis(False)
axes[1].set_ylabel("5m")
axes[2].imshow(cv2.imread(n2Rpath+n2_routefollow_[2]))
axes[2].axis(False)
axes[2].set_ylabel("10m")
axes[3].imshow(cv2.imread(n2Rpath+n2_routefollow_[3]))
axes[3].axis(False)
axes[3].set_ylabel("14M. Foraging tree")
plt.show()
```
