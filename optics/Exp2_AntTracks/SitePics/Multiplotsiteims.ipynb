```python
n1N = []
n1E = []
n1S = []
n1W = []
n1release = []

for file in os.listdir(n1Rpath):
    if file[-3:].lower() == 'jpg':
        print(file)
        if 'release' in file.lower():
            n1release.append(file)
        elif '_E' in file:
            n1E.append(file)
        elif '_W' in file:
            n1W.append(file)
        elif 'N2m' in file:
            n1N.append(file)
        elif '_S' in file:
            n1S.append(file)

print("release",len(n1release))
print("N",len(n1N))
print("E",len(n1E))
print("S",len(n1S))
print("W",len(n1W))
```

```python
n2_routefollow = []
for file in os.listdir(n2Rpath):
    if file[-3:].lower() == 'jpg':
        if 'onroute1' in file:
            n2_routefollow.append(file)
        elif 'base' in file:
            n2_routefollow.append(file)

print(len(n2_routefollow))
print(n2_routefollow)
#n2_routefollow_ = [n2_routefollow[0], n2_routefollow[2], n2_routefollow[1]]
n2_routefollow_ = [n2_routefollow[3], n2_routefollow[0], n2_routefollow[2], n2_routefollow[1]]

print(n2_routefollow_)
```
