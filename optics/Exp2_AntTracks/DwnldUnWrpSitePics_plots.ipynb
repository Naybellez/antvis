Read in the images. Unwrap. Resave in 'rect'

```python
def unwrap_allinpath(filepath, savepath):
    for impath in os.listdir(filepath):
        print(impath)
        if impath[-3:].lower() == "jpg":
            img = cv2.imread(filepath+impath)
            print("image read")
            unwrapped_img = Unwrap(img)
            print("image unwrapped")
            plt.imshow(unwrapped_img)
            #plt.title(f"{impath}")
            plt.axis(False)
            plt.tight_layout()
            filename = f"{savepath}{impath[:-4]}_unwrapped.jpg"
            #plt.savefig(f"{n1rpath}{impath[:-4]}_unwrapped.jpg")
            cv2.imwrite(filename, unwrapped_img)
            
```
