```python

def find_Var(df):
    df['ID'] = df['ID'].astype('str')
    
    ids = np.unique(df['ID'])
    #print(f"unique ids   {ids}")
    final_heading_list = []
    
    
    for id in ids:
        usedf = df[df['ID']== id]
        final_heading_list.append(np.array(usedf['theta'])[-1])
        
    final_heading_array = np.array(final_heading_list)
    #print(f"final heading array    {final_heading_array}")
    rayleigh_ans = astro.rayleightest(final_heading_array)
    print(f"rayleigh ans  {rayleigh_ans}")
    if rayleigh_ans <= 0.05:
        mean_heading_theta = scipy.stats.circmean(final_heading_array) # circ mean wants radians
        var_heading_theta = scipy.stats.circvar(final_heading_array)
        std_heading_theta = scipy.stats.circstd(final_heading_array)

        print(f"MEAN : {mean_heading_theta}     VARIENCE  {var_heading_theta}     STD  {std_heading_theta}     RADIANS")
        print(f"MEAN : {np.rad2deg(mean_heading_theta)}     VARIENCE  {np.rad2deg(var_heading_theta)}     STD  {np.rad2deg(std_heading_theta)}      DEGREES")
        print(f"mean - var  {np.rad2deg(mean_heading_theta - var_heading_theta)}")
        print(f"mean + var  {np.rad2deg(mean_heading_theta + var_heading_theta)}")
```
